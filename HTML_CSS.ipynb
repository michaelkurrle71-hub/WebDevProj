CSS stands for Cascading Style Sheets

## How to add CSS

[Example](./Abschnitt05/5.1.%20Adding%20CSS/)

```
Inline <tag style="css"/> 
    <html style="background: blue"></html>

Internal <style>css</style>
    <html>
        <head>
            <style>
                html{background: red;}
            </style>
        </head>
    </html>

External <link href="style.css"/> is in a external file like styles.css
    html{background: green;}

The external file has to be linked in the html file in the head section
<head>
    <link rel="stylesheet" href="./style.css"/>
</head>
```

## CSS Selectors

[Example](./Abschnitt05/5.3%20CSS%20Selectors/)

Element Selector
````
h1{h1 is a CSS Element Selector
    color: blue;
}
````

Class Selector: all HTML Elements which have the Class Selector gets the declaration
````
.red-text{dot no space
    color: red;
}
````
HTML-Code
````
<h3 class="red-text">red</h2>
<p class="red-text">I'm red too</p>
````

ID Selector it should be unique
````
#main{the Pound or Hash-Tag no space
    color: red;
}
HTML-Code
<p id="main>I'm a paragaph in red</p>
````

Hint Attributes Selector are for example src, alt, draggable<br/>
In a CSS file it can look like
````
p[draggable="false"]{
    color: red;
}

HTML-Code
<p draggable="false">Drag me I'm red</p>
<p draggable="true">Don't drag me because I'm not red</p>

Universal Selector
````
*{->select all is the asterix
    color: red;
}
````
